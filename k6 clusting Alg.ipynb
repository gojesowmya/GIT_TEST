```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
df_scaling = pd.DataFrame(sc.fit_transform(df),
                          columns=df.columns,
                          index=df.index)
```

## k-means

```python
## k-means->when we having the spatial data use k-means
## k-parameter
## firstly random k-points(k=2)
## from assigning data point to every point
## EUCIDEAN,MINKAWASKI, MANHATTON
## Based on nearest points to assign points
##moving point to center 
```

```python
from sklearn.cluster import KMeans
kmean = KMeans()
```

```python
kmean.fit(df_scaling)
```

```python
kmean.labels_
```

```python
df['kmean_label'] = kmean.labels_
```
